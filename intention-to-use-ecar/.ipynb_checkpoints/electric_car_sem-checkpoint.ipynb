```python
import pandas as pd
import numpy as np
import semopy
from semopy import Model
from statsmodels.stats.outliers_influence import variance_inflation_factor
from statsmodels.tools.tools import add_constant
```

```python
# Đọc file csv, tách dữ liệu để chạy MGA
df = pd.read_excel("Downloads/ecardata2503-373obs-1.xlsx")

df_male = df[df['GD'] == 1]
df_female = df[df['GD'] == 0]

print("Phần trăm nam giới trong mẫu:", df_male.shape[0]/df.shape[0]*100)
print("Phần trăm nữ giới trong mẫu:", df_female.shape[0]/df.shape[0]*100)
```

```python
# Kích thước và kiểu dữ liệu
print("Dataset info:")
df.info()
```

```python
# Thống kê mô tả
print("\nDescriptive statistics:")
display(df.describe())
```

```python
# Xác định model
model_desc = """
PEU =~ PEU1 + PEU2 + PEU3 + PEU4 + PEU5
PU =~ PU1 + PU2 + PU3 + PU4 + PU5 + PU6
FIP =~ FIP1 + FIP2 + FIP3
SJN =~ SJN1 + SJN2 + SJN3
PBC =~ PBC1 + PBC2 + PBC3 + PBC4 + PBC5 + PBC6 + PBC7
ATT =~ ATT1 + ATT2 + ATT3 + ATT4
ITU =~ ITU1 + ITU2 + ITU3 + ITU4
"""

model = Model(model_desc)
model.fit(df)
```

```python
# Lấy các biến quan sát đầu vào
latent_df = model.predict_factors(df)
latent_vars = ['PEU', 'PU', 'FIP','SJN','PBC', 'ATT','ITU']

# Kiểm tra VIF cho các predictors của ATT
predictors_att = ['PEU', 'PU']
X1 = add_constant(latent_df[predictors_att])

vif1 = pd.DataFrame()
vif1["Variable"] = X1.columns
vif1["VIF"] = [variance_inflation_factor(X1.values, i) for i in range(X1.shape[1])]

print("VIF giữa PEU và PU (ảnh hưởng ATT):")
print(vif1.sort_values(by="VIF", ascending=False))
print("\n")

# Kiểm tra VIF cho các predictors của ITU
predictors_itu = ['ATT', 'PBC', 'FIP', 'SJN']
X2 = add_constant(latent_df[predictors_itu])

vif2 = pd.DataFrame()
vif2["Variable"] = X2.columns
vif2["VIF"] = [variance_inflation_factor(X2.values, i) for i in range(X2.shape[1])]

print("VIF giữa ATT, PBC, FIP, SJN (ảnh hưởng ITU):")
print(vif2.sort_values(by="VIF", ascending=False))
```

```python
# Kiểm tra factor loading
estimates = model.inspect(std_est=True)
factor_loadings = estimates[estimates["op"] == "~"][["lval", "rval", "Est. Std"]]
print(factor_loadings)
```

```python
# Xác định model sau kiểm tra factor loading
model_desc = """
PEU =~ PEU1 + PEU2 + PEU3 + PEU4 + PEU5
PU =~ PU1 + PU2 + PU3 + PU4 + PU5
FIP =~ FIP1 + FIP2 + FIP3
SJN =~ SJN1 + SJN2 + SJN3
PBC =~ PBC1 + PBC3 + PBC4 + PBC5 + PBC6 + PBC7
ATT =~ ATT1 + ATT2 + ATT3 + ATT4
ITU =~ ITU1 + ITU2 + ITU3 + ITU4
ATT ~ PEU
ATT ~ PU
ITU ~ PEU
ITU ~ PU
ITU ~ ATT
ITU ~ SJN
ITU ~ PBC
ITU ~ FIP
"""

model = Model(model_desc)
```

```python
# Chạy mô hình SEM
model = semopy.Model(model_desc)
model.fit(df)

# In kết quả
print(model.inspect().head(8))
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

latent_df = model.predict_factors(df).dropna()

# Kiểm tra R-squared cho mô hình ATT ~ PU + PEU
X1 = latent_df[['PU', 'PEU']]
y1 = latent_df['ATT']

reg1 = LinearRegression().fit(X1, y1)
r2_att = r2_score(y1, reg1.predict(X1))
print(f"R² cho ATT: {r2_att:.4f}")

# Kiểm tra R-squared cho mô hình ITU ~ ATT + PBC + FIP + SJN
X2 = latent_df[['ATT', 'PBC', 'FIP', 'SJN']]
y2 = latent_df['ITU']

reg2 = LinearRegression().fit(X2, y2)
r2_itu = r2_score(y2, reg2.predict(X2))
print(f"R² cho ITU: {r2_itu:.4f}")
```

```python
# Fit model cho từng nhóm
model_male = Model(model_desc)
model_male.fit(df_male)
results_male = model_male.inspect(std_est=True)

model_female = Model(model_desc)
model_female.fit(df_female)
results_female = model_female.inspect(std_est=True)

latent_vars = ['PEU', 'PU', 'FIP', 'SJN', 'PBC', 'ATT', 'ITU']

# Lọc hệ số và SE của inner model
def get_inner_paths(data, label):
    model = Model(model_desc)
    model.fit(data)
    results = model.inspect(std_est=True)
    inner = results[
        (results['op'] == '~') &
        (results['lval'].isin(latent_vars)) &
        (results['rval'].isin(latent_vars))
    ][['lval', 'rval', 'Est. Std', 'p-value']].rename(columns={
        'Est. Std': f'Coef_{label}',
        'p-value': f'p_{label}'
    })
    return inner

# Lấy kết quả cho từng nhóm 
inner_male = get_inner_paths(df_male, 'Male')
inner_female = get_inner_paths(df_female, 'Female')

merged = pd.merge(inner_male, inner_female, on=['lval', 'rval'], how='outer')
merged['Difference'] = merged['Coef_Male'] - merged['Coef_Female']

# In kết quả
print("Multigroup Analysis:")
print(merged)
```
